```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

# utility function to plot functions
def plot(name: str, x: torch.Tensor, y: torch.Tensor, square: bool = True):
    data = {
        "x": x.tolist(),
        "y": y.tolist(),
        "c": (x * 0).tolist()
    }
    df = pd.DataFrame(data)
    
    sns.set_style("whitegrid")
    sns.lineplot(data=data, x="x", y="y", label=name).set_title(name)
    sns.lineplot(data=df, x="x", y="x", label="y = x", alpha=0.3, linestyle="--")
    sns.lineplot(data=df, x="x", y="c", label="y = 0", alpha=0.3, linestyle="--")

    if square:
        plt.xlim(min(data["x"]), max(data["x"]))
        plt.ylim(min(data["x"]), max(data["x"]))
        plt.yticks(range(int(min(data["x"])), int(max(data["x"])) + 1))
    else:
        plt.xlim(min(data["x"]), max(data["x"]))
        plt.ylim(min(data["y"]), max(data["y"]))
        plt.yticks(torch.linspace(float(min(data["y"])),float(max(data["y"])),10))

    plt.xticks(range(int(min(data["x"])), int(max(data["x"])) + 1))


    plt.show()
```

```python
from torch.nn import functional as F
import torch

x = torch.linspace(-10, 10, 100)
x
```

```python
relu = F.relu(x)
print(relu)
plot("ReLU", x, relu)
```

```python
gelu = F.gelu(x)
print(gelu)
plot("GELU", x, gelu)
plot("GELU", x, gelu, False)
```

```python
elu = F.elu(x)
print(elu)
plot("ELU", x, elu)
```

```python
celu = F.celu(x)
print(celu)
plot("CELU", x, celu)
```

```python
selu = F.selu(x)
print(selu)
plot("SELU", x, selu)
```

```python
silu = F.silu(x)
print(silu)
plot("SILU", x, silu)
```

```python
leaky_relu = F.leaky_relu(x)
print(leaky_relu)
plot("Leaky Relu", x, leaky_relu, False)
```

```python
softmax = F.softmax(x)
print(softmax)
#print(glu.shape, x.shape)
plot("Softmax", x, softmax)
plot("Softmax", x, softmax, False)
```

```python
sigmoid = F.sigmoid(x)
print(sigmoid)
plot("Sigmoid", x, sigmoid, False)
```

```python
tanh = F.tanh(x)
print(tanh)
plot("Tanh", x, tanh, False)
```
